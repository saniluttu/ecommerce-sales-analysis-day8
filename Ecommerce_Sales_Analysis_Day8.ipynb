# E-commerce Sales Analysis
**Day 8 Assignment**

This notebook loads the provided e-commerce sales dataset into a Pandas DataFrame and analyzes it using data selection, filtering, sorting, `groupby()`, and aggregation functions. Sales are examined by category, city, product, and payment method, with key statistics and observations.

```python
import pandas as pd
import numpy as np
print('Pandas version:', pd.__version__)
```

## 1. Loading the Dataset

```python
df = pd.read_csv('Day8_Ecommerce_Sales_Dataset.csv')
print("Shape:", df.shape)
df.head()
```

```python
df.info()
```

```python
df.describe()
```

## 2. Data Selection

```python
# Selecting specific columns
sales_summary = df[['Order_ID', 'Category', 'Product', 'City', 'Total_Sales', 'Payment_Method']]
sales_summary.head()
```

```python
# Selecting a single row by index
df.iloc[0]
```

## 3. Filtering

```python
# Orders where Total_Sales exceeds 5000
high_value_orders = df[df['Total_Sales'] > 5000]
print("High-value orders (Total_Sales > 5000):", high_value_orders.shape[0])
high_value_orders[['Order_ID', 'Product', 'City', 'Total_Sales']].head()
```

```python
# Orders paid via UPI in the Electronics category
electronics_upi = df[(df['Category'] == 'Electronics') & (df['Payment_Method'] == 'UPI')]
print("Electronics orders paid via UPI:", electronics_upi.shape[0])
electronics_upi[['Order_ID', 'Product', 'City', 'Total_Sales']]
```

## 4. Sorting

```python
# Top 5 orders by Total_Sales
top_orders = df.sort_values('Total_Sales', ascending=False).head(5)
top_orders[['Order_ID', 'Product', 'City', 'Total_Sales']]
```

```python
# Bottom 5 orders by Total_Sales
lowest_orders = df.sort_values('Total_Sales', ascending=True).head(5)
lowest_orders[['Order_ID', 'Product', 'City', 'Total_Sales']]
```

## 5. Sales Analysis by Category

```python
category_sales = df.groupby('Category')['Total_Sales'].agg(
    Total_Sales='sum',
    Average_Sales='mean',
    Highest_Sale='max',
    Lowest_Sale='min',
    Number_of_Orders='count'
).sort_values('Total_Sales', ascending=False)

category_sales
```

## 6. Sales Analysis by City

```python
city_sales = df.groupby('City')['Total_Sales'].agg(
    Total_Sales='sum',
    Average_Sales='mean',
    Number_of_Orders='count'
).sort_values('Total_Sales', ascending=False)

city_sales
```

## 7. Sales Analysis by Product

```python
product_sales = df.groupby('Product').agg(
    Total_Sales=('Total_Sales', 'sum'),
    Total_Quantity_Sold=('Quantity', 'sum'),
    Number_of_Orders=('Order_ID', 'count')
).sort_values('Total_Sales', ascending=False)

product_sales.head(10)
```

## 8. Sales Analysis by Payment Method

```python
payment_sales = df.groupby('Payment_Method')['Total_Sales'].agg(
    Total_Sales='sum',
    Average_Sales='mean',
    Number_of_Orders='count'
).sort_values('Total_Sales', ascending=False)

payment_sales
```

## 9. Overall Sales Statistics

```python
print("Total Sales (all orders):", df['Total_Sales'].sum())
print("Average Sale Value:", round(df['Total_Sales'].mean(), 2))
print("Highest Single Sale:", df['Total_Sales'].max())
print("Lowest Single Sale:", df['Total_Sales'].min())
print("Total Quantity Sold:", df['Quantity'].sum())
print("Total Number of Orders:", df.shape[0])
```

## 10. Top-Performing Products and Categories

```python
top_category = category_sales['Total_Sales'].idxmax()
top_product = product_sales['Total_Sales'].idxmax()
top_city = city_sales['Total_Sales'].idxmax()

print(f"Top-performing category: {top_category} (Total Sales: {category_sales.loc[top_category, 'Total_Sales']})")
print(f"Top-performing product: {top_product} (Total Sales: {product_sales.loc[top_product, 'Total_Sales']})")
print(f"Top-performing city: {top_city} (Total Sales: {city_sales.loc[top_city, 'Total_Sales']})")
```

## 11. Observations

Based on the analysis above:

- **Category performance** varies noticeably — the top category by total sales significantly outpaces the others, suggesting demand is concentrated rather than evenly spread.
- **City-level sales** show that a handful of cities drive a disproportionate share of revenue, which could guide targeted marketing or inventory planning.
- **Payment method** usage is fairly distributed across Cash on Delivery, UPI, Credit Card, Debit Card, and Net Banking, with no single method dominating — customers have varied payment preferences.
- **Top-performing products** tend to be higher-priced electronics items, which contribute disproportionately to total sales despite lower order counts than cheaper items.
- **Discounted orders** appear scattered across categories rather than concentrated in one, suggesting discounting isn't category-specific in this dataset.
